## Create Sample Dataset

```python
import pandas as pd 
data = {    'Distance': [2, 5, 3, None, 4, 6, 8, 2, 5, 9, 5],   
            'Traffic': ['Low', 'High', 'Medium', 'High', 'Low',             
                        'Medium', 'High', 'Low', 'Medium', 'High', 'High'],   
            'Weather': ['Sunny', 'Rainy', 'Cloudy', 'Rainy', 'Sunny',                'Cloudy', 'Sunny', 'Cloudy', 'Rainy', 'Sunny', 'High'],   
            'Vehicle': ['Bike', 'Bike', 'Scooter', 'Bike', 'Scooter',                'Bike', 'Scooter', 'Bike', 'Scooter', 'Bike', 'Bike'],   
            'Delivery_Time': [15, 35, 25, 45, None, 30, 50, 18, 32, 48, 35] }
df = pd.DataFrame(data) 
(df)
```

```python
df.isnull().sum()
```

```python
df['Distance'] = df['Distance'].fillna(df['Distance'].mean()) 
df['Delivery_Time'] = df['Delivery_Time'].fillna(  
    df['Delivery_Time'].mean() 
) 
(df) 
```

```python
print(df.duplicated().sum())
```
